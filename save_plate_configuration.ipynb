```python
import mdtraj as md
import numpy as np
```

```python

t = 300

ints = list(range(60, 151, 2))

distances_in_nm = [f"{d/100:.2f}" for d in ints]
base_path = f"/gibbs/arghavan/plate_simulations/{t}K/"
output_path = f"/gibbs/arghavan/hp-results-pc/plate_configurations/"

for d in distances_in_nm:

    d_in_Angstrom = round(float(d) * 10,1)

    frame = md.load_frame(f"{base_path}/{d}/{t}K_{d}.nc",
                         top=f"{base_path}/{d}/topol_edited.prmtop",
                         index=0)
    # tpl = md.load_topology(f"{base_path}/{d}/topol_edited.prmtop")
    walls = frame.topology.select("resname WALL")
    frame.atom_slice(walls).save_pdb(f"{output_path}/{d_in_Angstrom}_plate_configuration.pdb")
    
```
