```python
import numpy as np
import pandas as pd

base_path = '/gibbs/arghavan/gist_hydrophobic_plates/gist_results/'

N_nbr_bulk_tip4p2005 = 5.16431
N_nbr_bulk_15A = 4.93729
V_vox = 0.125
c = 0.46
bulk_six_entropy = np.float64(355.9010488361641)
bulk_orient_entropy = np.float64(226.56541491482477)
bulk_trans_entropy = np.float64(147.02892627401144)

# Distances: 5.4, 5.6, ..., 15.0
ints = list(range(54, 151, 2))
distances_in_A = [f"{d/10:.1f}" for d in ints]


entropy_results = []

for d in distances_in_A:

    print(f"Processing d = {d} Å")

    filtered_file = f'{base_path}{d}/d-{d}-output-filtered.dat'

    # Read the data
    df_filtered = pd.read_csv(
        filtered_file,
        sep=r'\s+',
        header=1,
        comment='#'
    )
    
    scale = 1.0 - c * (df_filtered['neighbor-norm'] / N_nbr_bulk_tip4p2005)
    scale = scale.clip(lower=0.0, upper=1.0)   # optional safety

    total_dTSsix_corrected_unref = (
        (df_filtered['dTSsix-dens(kcal/mol/A^3)'] * scale).sum() * (-V_vox)
    )
    total_dTSsix_corrected_ref = total_dTSsix_corrected_unref - bulk_six_entropy


    total_dTSorient_corrected_unref = (
        (df_filtered['dTSorient-dens(kcal/mol/A^3)'] * scale).sum() * (-V_vox)
    )   
    total_dTSorient_corrected_ref = total_dTSorient_corrected_unref - bulk_orient_entropy


    total_dTStrans_corrected_unref = (
        (df_filtered['dTStrans-dens(kcal/mol/A^3)'] * scale).sum() * (-V_vox)
    )
    total_dTStrans_corrected_ref = total_dTStrans_corrected_unref - bulk_trans_entropy


    # Orient + trans
    total_entropy_corr_ref = (
        total_dTSorient_corrected_ref + total_dTStrans_corrected_ref
    )



    # Save d and entropy to results
    entropy_results.append([float(d), total_dTSsix_corrected_ref, total_dTSorient_corrected_unref, total_dTStrans_corrected_unref, total_dTStrans_corrected_ref])




# Create the final results file
results_file = f'{base_path}/corrected_neg_dTS_0.dat'

with open(results_file, 'w') as f:
    f.write("distance(A)\t-dTSsix(ref)\t-dTSorient(ref)\t-dTStrans(ref)\t-dTSorient+trans(ref)\n")

    for d, *entropies in entropy_results:
        f.write(f"{d:.1f}\t" + "\t".join(f"{e:.8f}" for e in entropies) + "\n")

print(f"\nResults saved to: {results_file}")
```

```python
import numpy as np
import pandas as pd

base_path = '/gibbs/arghavan/gist_hydrophobic_plates/gist_results/'

N_nbr_bulk_tip4p2005 = 5.16431
N_nbr_bulk_15A = 4.93729

# bulk_entropy = np.float64(401.80649978375016)
# bulk_orient_entropy = np.float64(251.75385383375004)
# bulk_trans_entropy =  np.float64(167.5696186525)



entropy_results = []

d = '15.0'
filtered_file = f'{base_path}{d}/d-{d}-output-filtered.dat'

# Read the data
df_filtered = pd.read_csv(
    filtered_file,
    sep=r'\s+',
    header=1,
    comment='#'
)

# df_filtered['dTSsix_unref'] = df_filtered['dTSsix-dens(kcal/mol/A^3)'] * (-0.125)
# df_filtered['dTsorient_unref'] = df_filtered['dTSorient-dens(kcal/mol/A^3)'] * (-0.125)
# df_filtered['dTStrans_unref'] = df_filtered['dTStrans-dens(kcal/mol/A^3)'] * (-0.125)

df_filtered['dTSsix-dens-corrected'] = df_filtered['dTSsix-dens(kcal/mol/A^3)'] - 4.2 * df_filtered['neighbor-dens(1/A^3)'] / N_nbr_bulk_tip4p2005
df_filtered['dTStrans-dens-corrected'] = df_filtered['dTSorient-dens(kcal/mol/A^3)'] - 4.2 * df_filtered['neighbor-dens(1/A^3)']/ N_nbr_bulk_tip4p2005
df_filtered['dTSorient-dens-corrected'] = df_filtered['dTStrans-dens(kcal/mol/A^3)'] - 4.2 * df_filtered['neighbor-dens(1/A^3)'] / N_nbr_bulk_tip4p2005


# Six entropy
total_dTS_unref = (
    df_filtered['dTSsix-dens-corrected'].sum() * (-0.125)
)


# Orientational entropy
total_orient_entropy_unref = (
    df_filtered['dTSorient-dens-corrected'].sum() * (-0.125)
)


# Translational entropy
total_trans_entropy_unref = (
    df_filtered['dTStrans-dens-corrected'].sum() * (-0.125)
)


# Orient + trans
total_entropy_unref = (
    total_orient_entropy_unref + total_trans_entropy_unref
)



total_dTS_unref, total_orient_entropy_unref, total_trans_entropy_unref, total_entropy_unref
```
